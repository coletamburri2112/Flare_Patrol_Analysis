```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
CaII_met = np.load('/Users/coletamburri/Desktop/DKIST_August2024_Flare_RefResponse/11Aug2024_kmeans_SeptemberResults/CaIIH_kmeans_metrics.npz')
Hbeta_met = np.load('/Users/coletamburri/Desktop/DKIST_August2024_Flare_RefResponse/11Aug2024_kmeans_SeptemberResults/Hbeta_kmeans_metrics.npz')
```

```python
silhs_CaIIH = CaII_met['silhs']
dbinds_CaIIH = CaII_met['dbinds']
inertias_CaIIH = CaII_met['inertias']

silhs_Hb = Hbeta_met['silhs']
dbinds_Hb = Hbeta_met['dbinds']
inertias_Hb = Hbeta_met['inertias']
```

```python
np.shape(np.arange(2,49))
```

```python
fig,ax=plt.subplots(2,3,dpi=200,figsize=(12,5))

ax.flatten()[0].plot(np.arange(3,50),silhs_CaIIH,label='Ca II H',c='#CC6677',marker='.')
ax.flatten()[3].plot(np.arange(3,50),silhs_Hb,label=r'H$\beta$',c='#CC6677',marker='.')

ax.flatten()[1].plot(np.arange(3,50),dbinds_CaIIH,label='Ca II H',c='#CC6677',marker='.')
ax.flatten()[4].plot(np.arange(3,50),dbinds_Hb,label=r'H$\beta$',c='#CC6677',marker='.')

ax.flatten()[2].plot(np.arange(3,50),inertias_CaIIH/1e5,label='Ca II H',c='#CC6677',marker='.')
ax.flatten()[5].plot(np.arange(3,50),inertias_Hb/1e5,label=r'H$\beta$',c='#CC6677',marker='.')

ax.flatten()[0].axvline(11,c='black',linestyle='dashed')
ax.flatten()[1].axvline(11,c='black',linestyle='dashed')
ax.flatten()[2].axvline(11,c='black',linestyle='dashed')

ax.flatten()[0].axvline(12,c='black',linestyle='dotted',alpha=0.3)
ax.flatten()[1].axvline(12,c='black',linestyle='dotted',alpha=0.3)
ax.flatten()[2].axvline(12,c='black',linestyle='dotted',alpha=0.3)

ax.flatten()[3].axvline(14,c='black',linestyle='dashed')
ax.flatten()[4].axvline(14,c='black',linestyle='dashed')
ax.flatten()[5].axvline(14,c='black',linestyle='dashed')

ax.flatten()[0].grid()
ax.flatten()[1].grid()
ax.flatten()[2].grid()
ax.flatten()[3].grid()
ax.flatten()[4].grid()
ax.flatten()[5].grid()

ax.flatten()[0].set_ylabel('Silhouette Score')
ax.flatten()[1].set_ylabel('Davies-Bouldin Index')
ax.flatten()[2].set_ylabel(r'Inertia [x $10^5$]')

ax.flatten()[3].set_ylabel('Silhouette Score')
ax.flatten()[4].set_ylabel('Davies-Bouldin Index')
ax.flatten()[5].set_ylabel(r'Inertia [x $10^5$]')

ax.flatten()[0].set_xlabel(r'$k$')
ax.flatten()[1].set_xlabel(r'$k$')
ax.flatten()[2].set_xlabel(r'$k$')

ax.flatten()[3].set_xlabel(r'$k$')
ax.flatten()[4].set_xlabel(r'$k$')
ax.flatten()[5].set_xlabel(r'$k$')

fig.tight_layout()
```
